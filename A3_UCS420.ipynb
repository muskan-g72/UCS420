```python
#Question 6
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
employees = pd.DataFrame(employee_data)
employees.to_csv("employees.csv", index=False)
print(employees)
```

```python
print("Shape of DataFrame:", employees.shape)
print("Number of rows:", employees.shape[0])
print("Number of columns:", employees.shape[1])
```

```python
employees.info()
```

```python
print(employees.describe())
```

```python
print("First five rows:")
print(employees.head(5))
print("\nLast three rows:")
print(employees.tail(3))
```

```python
print("Average salary:", employees["Salary"].mean())
print("Total bonus:", employees["Bonus"].sum())
print("Youngest age:", employees["Age"].min())
print("Highest rating:", employees["Rating"].max())
```

```python
print("Salary in descending order:")
salary_sorted = employees.sort_values(
    by="Salary",
    ascending=False
)
print(salary_sorted)
```

```python
def categorize_performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"
employees["Performance_Category"] = employees["Rating"].apply(categorize_performance)
print("Performance categories:")
print(employees[["Name", "Rating", "Performance_Category"]])
```

```python
print("Missing values:")
print(employees.isnull().sum())
```

```python
employees.rename(columns={"Employee_ID": "ID"},inplace=True)
print(" Columns after renaming:")
print(employees.columns)
```

```python

print(employees[employees["Years_of_Experience"] > 5])
print(employees[employees["Department"] == "IT"])
```

```python
employees["Tax"] = employees["Salary"] * 0.10
print("Tax details:")
print(employees[["Name", "Salary", "Tax"]])
```

```python
employees.to_csv("modified_employees.csv",index=False)
print("File saved as modified_employees.csv")
```
