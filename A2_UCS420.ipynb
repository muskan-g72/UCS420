```python
#question 1
#(i)Print L
roll_no="1024160074"
digits=[int(digit) for digit in roll_no]
L=[digit*10 for digit in digits]
print("Initial list L")
print(L)
```

```python
#(ii)Add two numbers of your choice to L
L.append(50)
print(f"L after append(50):{L}")
#append() adds new element at the end of list.
L.insert(3,30)
print(f"L after insert(3,30):{L}")
#insert() adds element at specified index and shifts later elements
```

```python
#(iii)Remove two elements from L
L.remove(10)
print(f"L after remove(10): {L}")
ele=L.pop()
print(f"L after pop():{L}")
print(f"removed element : {ele}")
```

```python
#(iv) Sort L
L.sort()
print(f"L in ascending order:{L}")
L.sort(reverse=True)
print(f"L in descending order:{L}")
```

```python
#(v) Print first and last three elements
print(f"first three elements:{L[:3]}")
print(f"last three elements:{L[-3:]}")
```

```python
#(vi)Elements greater than average
avg=sum(L)/len(L)
greater_elements=[value for value in L if value>avg]
print(f"Average of L:{avg}")
print(f"Elements greater than average:{greater_elements}")
```

```python
#Question 3
#Generate a list of 100 random numbers
import random
from collections import Counter
random.seed(roll_no)
random_nos=[random.randint(100,900) for _ in range(100)]
print(f"Random Numbers:{random_nos}")
```

```python
# Count and print all odd numbers
odd_nos=[number for number in random_nos if number%2!=0]
print(f"Odd numbers:{odd_nos}")
print (f"Number of odd numbers:{len(odd_nos)}")
```

```python
# (iii) Count and print all even numbers
even_nos=[number for number in random_nos if number%2==0]
print(f"Even numbers:{even_nos}")
print (f"Number of even numbers:{len(even_nos)}")
```

```python
#(iv) Find prime Numbers
def is_prime(number):
  if number<2:
    return False
  for i in range(2,int(number**0.5)+1):
    if number%i==0:
      return False
    return True
prime_nos=[number for number in random_nos if is_prime(number)]
print(f"Prime Numbers:{prime_nos}")
print(f"Number of prime numbers:{len(prime_nos)}")
```

```python
#(v) Most frequently occurring number
frequency=Counter(random_nos)
most_frequent_no,occurrence = frequency.most_common(1)[0]
print(f"Most frequently occurring Number : {most_frequent_no} with count of :{occurrence}")
```

```python
#Question 5
# (i) Rename city to location using pop()
my_dict = {
    "name": "Muskan",
    "roll_no": "1024160074",
    "branch": "CSE",
    "age": int(input("\nEnter your current age: ")),
    "city": "Jhajjar"
}
print(f"Original Dictionary: {my_dict}")
my_dict["location"] = my_dict.pop("city")
print(f"After renaming 'city' to 'location':{my_dict}")
```

```python
# (ii) Add CGPA
my_dict["cgpa"] = 9.47
print(f"After adding CGPA:{my_dict}")
```

```python
# (iii) Increase age by 1
my_dict["age"] = my_dict["age"] + 1
print(f"After increasing age by 1:{my_dict}")
```

```python
# (iv) Delete branch using pop() and del in separate copies
dict_using_pop = my_dict.copy()
dict_using_del = my_dict.copy()
removed_branch = dict_using_pop.pop("branch")
print(f"Dictionary after deleting branch using pop():{dict_using_pop}")
print("Value returned by pop():", removed_branch)
del dict_using_del["branch"]
print(f"Dictionary after deleting branch using del:{dict_using_del}")
# pop() removes a key and returns its value, while del removes the key but does not return its value.

```

```python
# (v) Iterate using .items()
print("Key-value pairs:")
for key, value in my_dict.items():
    print(key, "→", value)
```

```python
# (vi) Safely check for email
print("Checking for email:")
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist in the dictionary.")
```

```python
# (vii) Create friend_dict and merge dictionaries
friend_dict = {
    "name": "Zaira",
    "roll_no": "1024160093",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}
merged_dict = {**my_dict, **friend_dict}

print(f"Friend Dictionary:{friend_dict}")
print(f"Merged Dictionary:{merged_dict}")
# If both dictionaries contain the same key, the value from the dictionary appearing later wins.
```

```python
# (viii) Dictionary containing only string values
string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}
print(f"Dictionary containing only string values:{string_values_dict}")
```
