```python
# Ques 1
import re
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.probability import FreqDist
from nltk.corpus import stopwords

text = """
Artificial intelligence is transforming the world of technology.
It helps computers learn from data and solve complex problems.
Machine learning is used in healthcare, education, and business.
I enjoy learning about AI because it makes everyday tasks easier.
In the future, intelligent systems may improve many aspects of life.
"""

# 1. Convert to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("Cleaned Text:\n", clean_text)

# 2. Tokenize into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:\n", words)
print("\nSentence Tokens:\n", sentences)

# 3. Compare split() and word_tokenize()
print("\nUsing split():\n", clean_text.split())
print("\nUsing word_tokenize():\n", word_tokenize(clean_text))

# 4. Remove stopwords
filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]
print("\nWords after removing stopwords:\n", filtered_words)

# 5. Word frequency distribution
freq_dist = FreqDist(filtered_words)
print("\nWord Frequencies:")
for word, frequency in freq_dist.most_common():
    print(word, ":", frequency)

# Plot word frequencies
freq_dist.plot(10, title="Top 10 Word Frequencies")
plt.show()
```

```python
#Ques 2
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = """
Artificial intelligence is transforming the world of technology.
It helps computers learn from data and solve complex problems.
Machine learning is used in healthcare, education, and business.
I enjoy learning about AI because it makes everyday tasks easier.
In the future, intelligent systems may improve many aspects of life.
"""

# 1. Extract words containing only alphabets
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print("Alphabetic Words:\n", words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [w for w in words if w not in stop_words]
print("\nAfter Stopword Removal:\n", filtered_words)

# 3. Perform stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(w) for w in filtered_words]
print("\nStemmed Words:\n", stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(w) for w in filtered_words
]
print("\nLemmatized Words:\n", lemmatized_words)

# 5. Compare outputs
print("\nComparison:")
for original, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:15} Stem: {stem:12} Lemma: {lemma}")
```

```python
# Ques 3
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Three short texts
texts = [
    "Artificial intelligence improves healthcare services.",
    "Machine learning helps computers analyze data.",
    "Healthcare technology uses artificial intelligence."
]

# 1. Bag of Words
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

# 2. TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray().round(3))

# 3. Top 3 keywords from each text
print("\nTop 3 Keywords for Each Text:")

for i, row in enumerate(tfidf_matrix.toarray()):
    top_indices = row.argsort()[::-1][:3]
    top_words = [
        (feature_names[j], round(row[j], 3))
        for j in top_indices if row[j] > 0
    ]
    print(f"Text {i+1}: {top_words}")
```

```python
# Ques 4
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Technology 1: Artificial Intelligence
ai_text = """
Artificial intelligence enables machines to learn from data.
It helps computers recognize patterns and solve complex problems.
AI is used in healthcare, education, and transportation.
It can automate repetitive tasks and support decision making.
AI systems improve through training and experience.
"""

# Technology 2: Blockchain
blockchain_text = """
Blockchain stores information in a distributed digital ledger.
It uses cryptography to protect transactions and records.
Blockchain supports secure payments and transparent data sharing.
It can be used in finance, supply chains, and digital identity.
Blockchain networks validate transactions through consensus.
"""

# 1. Preprocess and tokenize
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    tokens = word_tokenize(text)
    return [
        word for word in tokens
        if word not in stop_words
    ]

ai_words = preprocess(ai_text)
blockchain_words = preprocess(blockchain_text)

print("AI Tokens:\n", ai_words)
print("\nBlockchain Tokens:\n", blockchain_words)

# 2a. Jaccard similarity
set_ai = set(ai_words)
set_blockchain = set(blockchain_words)

intersection = set_ai.intersection(set_blockchain)
union = set_ai.union(set_blockchain)

jaccard = len(intersection) / len(union)

print("\nCommon Words:", intersection)
print("Jaccard Similarity:", round(jaccard, 4))

# 2b. Cosine similarity using TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([ai_text, blockchain_text])

cosine = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine, 4))
```

```python
# Ques 5
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Product reviews
reviews = [
    "This phone has an amazing camera and excellent battery life.",
    "I love this product because it is fast and easy to use.",
    "The quality is terrible and the battery drains quickly.",
    "The product is disappointing and the customer service is poor.",
    "The phone is okay and works as expected."
]

positive_reviews = []

print("Sentiment Analysis Results:\n")

# 1. Find polarity and subjectivity
for i, review in enumerate(reviews, start=1):
    analysis = TextBlob(review).sentiment
    polarity = analysis.polarity
    subjectivity = analysis.subjectivity

    # 2. Classify sentiment
    if polarity > 0.1:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < -0.1:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(f"Review {i}: {review}")
    print(f"Polarity: {polarity:.3f}")
    print(f"Subjectivity: {subjectivity:.3f}")
    print(f"Sentiment: {sentiment}\n")

# 3. Create word cloud for positive reviews
if positive_reviews:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white",
        stopwords=set(['this', 'and', 'it', 'is'])
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
```

```python
#Ques 6
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Training paragraph (~100 words)
text = """
Technology is changing the way people learn and communicate.
Students use computers to explore new ideas and solve difficult problems.
Artificial intelligence helps machines learn from data and recognize patterns.
Machine learning allows computers to improve their predictions through experience.
Deep learning uses neural networks to process complex information.
These technologies support healthcare, education, agriculture, and transportation.
Developers build intelligent applications to make everyday tasks easier.
Data plays an important role in training reliable machine learning models.
Good data helps a model discover useful patterns and make better predictions.
Researchers continue to develop safer and more efficient intelligent systems.
In the future, technology may help people solve many important challenges.
Learning programming helps students understand how these systems work.
Practice and experimentation are essential for building useful applications.
"""

# 2. Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1
print("Vocabulary Size:", total_words)

# Create input sequences
input_sequences = []

for sentence in text.lower().split('.'):
    sequence = tokenizer.texts_to_sequences([sentence])[0]

    for i in range(2, len(sequence) + 1):
        input_sequences.append(sequence[:i])

# Pad sequences
max_sequence_len = max(len(seq) for seq in input_sequences)

padded_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

# Input: all words except the last
# Target: the last word
X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

print("Input Shape:", X.shape)
print("Target Shape:", y.shape)

# 3. Build an LSTM model
model = Sequential([
    Embedding(total_words, 32),
    LSTM(64),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()

# 4. Train the model
model.fit(
    X, y,
    epochs=150,
    verbose=0
)

print("\nModel training completed!")

# 5. Generate text from a seed word
def generate_text(seed_word, next_words=8):
    generated = seed_word.lower()

    for _ in range(next_words):
        sequence = tokenizer.texts_to_sequences([generated])[0]

        if not sequence:
            break

        sequence = pad_sequences(
            [sequence],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predictions = model.predict(sequence, verbose=0)
        predicted_id = int(np.argmax(predictions, axis=1)[0])

        if predicted_id == 0:
            break

        predicted_word = tokenizer.index_word.get(predicted_id)

        if predicted_word is None:
            break

        generated += " " + predicted_word

    return generated

# Generate 2-3 lines from seed words
print("\nGenerated Text:")
print("1.", generate_text("technology", 8))
print("2.", generate_text("learning", 8))
print("3.", generate_text("data", 8))
```
