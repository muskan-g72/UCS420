```python
# Question 1
import pandas as pd

roll_no = input("Enter your college roll number: ")
digits = [int(d) for d in roll_no[-2:]]

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

constructed_entries = []

for digit in digits:
    category = categories[digit % 3]

    if category == "billing":
        question = "how do i pay my bill"
        answer = "You can pay your bill using UPI, card, or net banking."
        keywords = "bill payment fee pay"

    elif category == "account":
        question = "how do i update my account"
        answer = "Go to Settings > Account to update your details."
        keywords = "account update details profile"

    else:
        question = "how can i contact support"
        answer = "You can contact support during working hours."
        keywords = "support help contact assistance"

    constructed_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

faq_data = fixed_entries + constructed_entries

df = pd.DataFrame(faq_data)

print(df)
```

```python
# Question 2
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words.intersection(keywords))

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    return results
query = input("Enter your query: ")
results = score_query(query, df)
for result in results:
    print(result)
```

```python
# Question 3
def same_category(category_name, df):
    return df[df["category"].str.lower() == category_name.lower()]


category_name = input("Enter category: ")

print(same_category(category_name, df))
```

```python
# Question 4
index = 0

new_keyword = input("Enter a new keyword: ")

keywords = df.loc[index, "keywords"]

if new_keyword.lower() not in keywords.lower().split():
    df.loc[index, "keywords"] = keywords + " " + new_keyword

filename = roll_no + "_faq_data.csv"

df.to_csv(filename, index=False)

print(df)
print("Saved as:", filename)
```

```python
# Question 6
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words.intersection(keywords))

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    if results:
        highest_score = results[0]["score"]
        highest_matches = [
            result for result in results
            if result["score"] == highest_score
        ]

        print("Highest score:", highest_score)

        if len(highest_matches) > 1:
            print("Tie found:")
            for result in highest_matches:
                print(result)
        else:
            print("Best match:")
            print(highest_matches[0])
    else:
        print("No matching entries found.")


print("Tie query:")
score_query("fee", df)

print("\nNon-tie query:")
score_query("password reset", df)
```
